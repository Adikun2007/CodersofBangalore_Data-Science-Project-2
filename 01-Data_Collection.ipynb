```python
with open("finaldata.txt","r",encoding = 'utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
def parse_chunk(chunk):
    
    chunk = chunk.strip()
    separated_chunk = chunk.split("\n")
    user_name = separated_chunk[0]
    
    no_of_posts = int(separated_chunk[1].split(" post")[0].replace(",", ""))
    
    no_of_followers = float(separated_chunk[2].split(" follower")[0].replace(",", "").replace("K", "").replace("M", ""))
    if ("K" in separated_chunk[2]):
        no_of_followers = int(no_of_followers*1000)
    elif ("M" in separated_chunk[2]):
        no_of_followers = int(no_of_followers*1000000)
    else:
        no_of_followers=int(no_of_followers)
        
    no_of_following = float(separated_chunk[3].split(" following")[0].replace(",", "").replace("K", "").replace("M", ""))
    if ("K" in separated_chunk[3]):
        no_of_following = int(no_of_following*1000)
    elif ("M" in separated_chunk[3]):
        no_of_following = int(no_of_following*1000000)
    else:
        no_of_following=int(no_of_following)
    
    real_name = separated_chunk[4]
    if (len(separated_chunk)>5):
        type_of_page = separated_chunk[5]
        user_bio = "\n".join(separated_chunk[6:])
    else:
        type_of_page = "Unknown"
        user_bio = ""
        
    return {"user_name": user_name, "no_of_posts": no_of_posts, "no_of_followers": no_of_followers, 
            "no_of_following": no_of_following, "real_name": real_name, "type_of_page": type_of_page, 
            "user_bio": user_bio} 
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s = json.dumps(all_chunks, indent=4)
with open("Data.json", "w") as f:
    f.write(s)
```

# Who has the maximum posts
### finding out who has maximum posts

```python
max = 0
for chunk in all_chunks:
    if (max<chunk['no_of_posts']):
        max = chunk['no_of_posts']
        chunk_with_max_posts = chunk
print(chunk_with_max_posts['user_name'])
```

# who has the maximum followers

#### finding out who has maximum no of followers

```python
max_follower = 0
for chunk in all_chunks:
    if (max_follower < chunk['no_of_followers']):
        max_follower = chunk['no_of_followers']
        chunk_with_max_follower = chunk
print(chunk_with_max_follower['user_name'])
```

# Who follows the maximum people
#### finding out who follows maximum no of people

```python
max_following = 0
for chunk in all_chunks:
    if (max_following < chunk['no_of_following']):
        max_following = chunk['no_of_following']
        chunk_with_max_following = chunk
print(chunk_with_max_following['user_name'])
```
